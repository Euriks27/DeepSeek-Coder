````python
macro_engine_content = """
# macrolab/macro_engine.py

import os
import pandas as pd
import numpy as np
from datetime import datetime
import requests
from io import StringIO
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tsa.vector_ar.var_model import VAR
from linearmodels.panel import PanelOLS
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Assuming these are imported from other modules of macrolab, but for a single file, defining them here
# from .base_engine import DataLoader, DataPreprocessor
# from .econometrics import SPJEstimator, VARVECM, adf_test
# from .models import LSTMModel, KinkFiscalModel

# Placeholder implementations of DataLoader, DataPreprocessor, etc. for standalone operation if needed
# In a full package, these would be imported from their respective files.

class DataLoader:
    def __init__(self, file_path=None, data=None):
        self.file_path = file_path
        self.data = data

    def load_csv(self):
        if self.file_path:
            try:
                self.data = pd.read_csv(self.file_path)
                print(f"Data loaded successfully from {self.file_path}")
                return self.data
            except FileNotFoundError:
                print(f"Error: File not found at {self.file_path}")
                return None
        return self.data

    def load_bcb_series(self, series_code, start_date=None, end_date=None):
        url = f"https://api.bcb.gov.br/dados/serie/bcdata.sgs.{series_code}/dados?formato=json"
        if start_date:
            url += f"&dataInicial={start_date}"
        if end_date:
            url += f"&dataFinal={end_date}"

        response = requests.get(url)
        if response.status_code == 200:
            df = pd.DataFrame(response.json())
            df['data'] = pd.to_datetime(df['data'], format='%d/%m/%Y')
            df['valor'] = pd.to_numeric(df['valor'])
            df = df.set_index('data')
            self.data = df
            print(f"Series {series_code} loaded successfully from BCB.")
            return df
        else:
            print(f"Error loading series {series_code} from BCB: {response.status_code}")
            return None

    def get_info(self):
        if self.data is not None:
            print("\n--- Data Info ---")
            self.data.info()
            print("\n--- First 5 Rows ---")
            print(self.data.head())
        else:
            print("No data loaded yet.")


class DataPreprocessor:
    def __init__(self, dataframe):
        self.df = dataframe.copy() if dataframe is not None else pd.DataFrame()

    def handle_missing_values(self, strategy='mean', columns=None):
        if self.df.empty:
            print("DataFrame is empty, cannot preprocess.")
            return self.df

        cols_to_process = columns if columns else self.df.columns
        for col in cols_to_process:
            if self.df[col].isnull().any():
                if self.df[col].dtype in ["int64", "float64"]:
                    if strategy == 'mean':
                        self.df[col].fillna(self.df[col].mean(), inplace=True)
                    elif strategy == 'median':
                        self.df[col].fillna(self.df[col].median(), inplace=True)
                    elif strategy == 'zero':
                        self.df[col].fillna(0, inplace=True)
                    print(f"Missing values in column '{col}' filled with '{strategy}'")
                else:
                    print(f"Skipping non-numeric column '{col}' for missing value imputation.")
            else:
                print(f"No missing values found in column '{col}'")
        return self.df

    def normalize_data(self, columns=None, method='minmax'):
        if self.df.empty:
            print("DataFrame is empty, cannot normalize.")
            return self.df

        cols_to_normalize = columns if columns else self.df.select_dtypes(include=np.number).columns
        for col in cols_to_normalize:
            if method == 'minmax':
                scaler = MinMaxScaler()
            elif method == 'zscore':
                scaler = StandardScaler()
            else:
                print(f"Invalid normalization method: {method}")
                continue

            self.df[col] = scaler.fit_transform(self.df[[col]])
            print(f"Column '{col}' normalized using '{method}' method.")
        return self.df

    def aggregate_data(self, group_by_col, agg_col, agg_func='mean'):
        if self.df.empty:
            print("DataFrame is empty, cannot aggregate.")
            return self.df

        if group_by_col not in self.df.columns or agg_col not in self.df.columns:
            print("Specified group_by_col or agg_col not found in DataFrame.")
            return self.df

        print(f"Aggregating '{agg_col}' by '{group_by_col}' using '{agg_func}'")
        return self.df.groupby(group_by_col)[agg_col].agg(agg_func).reset_index()


class MacroLabEngineV10Complete:
    def __init__(self, project_root='.', data_source='csv', file_path=None, bcb_series_codes=None,
                 synthetic_data_params=None, start_date='2000-01-01', end_date='2026-12-31',
                 target_variable=None, features=None, country_col=None, time_col=None):

        self.project_root = project_root
        self.data_source = data_source
        self.file_path = file_path
        self.bcb_series_codes = bcb_series_codes if bcb_series_codes is not None else {}
        self.synthetic_data_params = synthetic_data_params if synthetic_data_params is not None else {}
        self.start_date = pd.to_datetime(start_date)
        self.end_date = pd.to_datetime(end_date)
        self.df = None
        self.target_variable = target_variable
        self.features = features
        self.country_col = country_col
        self.time_col = time_col
        self.model_results = {}
        self.scaler = None
        self.lstm_model = None

    def _generate_synthetic_data(self, params):
        print("Gerando dados sintéticos...")
        np.random.seed(42)
        num_samples = params.get('num_samples', 100)
        num_features = params.get('num_features', 3)
        feature_names = [f'Feature_{i+1}' for i in range(num_features)]
        data = np.random.randn(num_samples, num_features)
        df = pd.DataFrame(data, columns=feature_names)
        df['Target'] = df[feature_names[0]] * 2 + np.random.randn(num_samples) * 0.5
        if 'dates' in params:
            df['Date'] = pd.to_datetime(pd.date_range(start=params['dates']['start'], periods=num_samples, freq='M'))
            df = df.set_index('Date')
        print("Dados sintéticos gerados.")
        return df

    def ingest_data(self):
        print(f"Ingerindo dados de {self.data_source}...")
        if self.data_source == 'csv':
            loader = DataLoader(file_path=self.file_path)
            self.df = loader.load_csv()
        elif self.data_source == 'bcb_api':
            all_series_df = pd.DataFrame()
            for name, code in self.bcb_series_codes.items():
                loader = DataLoader()
                series_data = loader.load_bcb_series(code, self.start_date.strftime('%d/%m/%Y'), self.end_date.strftime('%d/%m/%Y'))
                if series_data is not None:
                    all_series_df[name] = series_data['valor']
            self.df = all_series_df
        elif self.data_source == 'synthetic':
            self.df = self._generate_synthetic_data(self.synthetic_data_params)
        else:
            raise ValueError("Fonte de dados inválida. Escolha entre 'csv', 'bcb_api' ou 'synthetic'.")

        if self.df is not None and not self.df.empty:
            print("Pré-processando dados ingeridos...")
            preprocessor = DataPreprocessor(self.df)
            self.df = preprocessor.handle_missing_values(strategy='mean') # Default strategy
            # self.df = preprocessor.normalize_data() # Normalizar dados pode ser feito antes do LSTM
            print("Dados ingeridos e pré-processados inicialmente.")
        else:
            print("Nenhum dado ingerido ou o DataFrame está vazio.")

    def perform_unit_root_tests(self, columns=None):
        print("Realizando testes de raiz unitária (ADF e KPSS)...")
        if self.df is None or self.df.empty:
            print("Nenhum dado disponível para testes de raiz unitária.")
            return

        cols = columns if columns else self.df.select_dtypes(include=np.number).columns
        results = {}
        for col in cols:
            if self.df[col].isnull().any():
                print(f"Skipping unit root test for {col} due to missing values.")
                continue
            try:
                # ADF Test
                adf_result = adfuller(self.df[col].dropna())
                adf_pvalue = adf_result[1]
                # KPSS Test
                kpss_result = kpss(self.df[col].dropna(), regression='c')
                kpss_pvalue = kpss_result[1]

                results[col] = {
                    'ADF_Statistic': adf_result[0],
                    'ADF_P-value': adf_pvalue,
                    'ADF_Stationary': 'Yes' if adf_pvalue <= 0.05 else 'No',
                    'KPSS_Statistic': kpss_result[0],
                    'KPSS_P-value': kpss_pvalue,
                    'KPSS_Stationary': 'Yes' if kpss_pvalue >= 0.05 else 'No' # KPSS null is stationary
                }
                print(f"Testes de raiz unitária para {col}: ADF p-value={adf_pvalue:.4f}, KPSS p-value={kpss_pvalue:.4f}")
            except Exception as e:
                print(f"Erro ao executar testes de raiz unitária para {col}: {e}")
        self.model_results['unit_root_tests'] = results
        return results

    def estimate_spj(self, dependent_var, independent_vars, entity_effect=True, time_effect=True):
        print("Estimando modelo SPJ (PanelOLS)...")
        if self.df is None or self.df.empty:
            print("Nenhum dado disponível para estimação SPJ.")
            return
        if self.country_col is None or self.time_col is None:
            print("Colunas de país e tempo são necessárias para PanelOLS.")
            return

        exog_vars = self.df[independent_vars].copy()
        exog_vars = sm.add_constant(exog_vars, prepend=False) # Add a constant
        dependent = self.df[dependent_var]

        # Create a PanelOLS friendly DataFrame
        panel_df = self.df.set_index([self.country_col, self.time_col])
        dependent_panel = panel_df[dependent_var]
        exog_panel = panel_df[exog_vars.columns]

        try:
            model = PanelOLS(dependent_panel, exog_panel, entity_effects=entity_effect, time_effects=time_effect)
            results = model.fit()
            self.model_results['spj_estimator'] = results
            print("Estimação SPJ concluída.\n")
            print(results)
            return results
        except Exception as e:
            print(f"Erro na estimação SPJ: {e}")
            return None

    def estimate_var_vecm(self, variables, lags=1, model_type='VAR'):
        print(f"Estimando modelo {model_type}...")
        if self.df is None or self.df.empty:
            print(f"Nenhum dado disponível para estimação {model_type}.")
            return

        data = self.df[variables].dropna()
        if data.empty:
            print(f"Dados vazios após remover NaNs para {model_type} com as variáveis {variables}.")
            return

        if model_type == 'VAR':
            try:
                model = VAR(data)
                results = model.fit(lags)
                self.model_results['var_vecm'] = results
                print(f"Estimação VAR com {lags} lags concluída.\n")
                print(results.summary())
                return results
            except Exception as e:
                print(f"Erro na estimação VAR: {e}")
                return None
        elif model_type == 'VECM':
            print("VECM não implementado ainda. Usando VAR como fallback.")
            # VECM implementation would require cointegration test first
            return self.estimate_var_vecm(variables, lags, model_type='VAR')
        else:
            raise ValueError("Tipo de modelo inválido. Escolha 'VAR' ou 'VECM'.")

    def mapeamento_kink_fiscal(self, tax_rate_series, threshold_series, income_series, tax_elasticity):
        print("Realizando mapeamento de kink fiscal...")
        if self.df is None or self.df.empty:
            print("Nenhum dado disponível para mapeamento de kink fiscal.")
            return

        # Ensure all series are aligned by index (e.g., Date)
        aligned_df = self.df[[tax_rate_series, threshold_series, income_series]].dropna()

        if aligned_df.empty:
            print("Dados insuficientes para mapeamento de kink fiscal após alinhamento e remoção de NaNs.")
            return None

        effective_tax = []
        for index, row in aligned_df.iterrows():
            tax_rate = row[tax_rate_series]
            threshold = row[threshold_series]
            income = row[income_series]

            if income <= threshold:
                effective_tax.append(income * tax_rate)
            else:
                # Simplified Kink logic: higher rate above threshold
                effective_tax.append(threshold * tax_rate + (income - threshold) * (tax_rate + tax_elasticity))

        self.df['Effective_Tax_Kink'] = pd.Series(effective_tax, index=aligned_df.index)
        print("Mapeamento de kink fiscal concluído. 'Effective_Tax_Kink' adicionado ao DataFrame.")
        return self.df

    def _create_sequences(self, data, seq_length):
        X, y = [], []
        for i in range(len(data) - seq_length):
            X.append(data[i:(i + seq_length)])
            y.append(data[i + seq_length])
        return np.array(X), np.array(y)

    def ancoragem_hibrida_lstm(self, target_variable, features, seq_length=10, epochs=50, batch_size=32,
                               validation_split=0.2, lstm_units=50, dropout_rate=0.2, patience=10):
        print("Iniciando ancoragem híbrida com LSTM...")
        if self.df is None or self.df.empty:
            print("Nenhum dado disponível para LSTM.")
            return

        # Prepare data
        data_for_lstm = self.df[[target_variable] + features].copy()
        data_for_lstm = data_for_lstm.dropna()

        if data_for_lstm.empty:
            print("Dados insuficientes para LSTM após remover NaNs.")
            return

        # Normalize data (MinMaxScaler is often preferred for LSTMs)
        self.scaler = MinMaxScaler(feature_range=(0, 1))
        scaled_data = self.scaler.fit_transform(data_for_lstm)

        # Create sequences
        X, y = self._create_sequences(scaled_data, seq_length)

        if X.shape[0] == 0:
            print("Não há sequências suficientes para criar o modelo LSTM. Ajuste 'seq_length'.")
            return

        # Reshape for LSTM [samples, time_steps, features]
        X = X.reshape(X.shape[0], X.shape[1], X.shape[2])

        # Split data
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=validation_split, random_state=42, shuffle=False)

        # Build LSTM Model
        self.lstm_model = Sequential([
            LSTM(lstm_units, activation='relu', input_shape=(X_train.shape[1], X_train.shape[2]), return_sequences=True),
            Dropout(dropout_rate),
            LSTM(lstm_units, activation='relu', return_sequences=False),
            Dropout(dropout_rate),
            Dense(y_train.shape[1]) # Output layer should match number of target features
        ])
        self.lstm_model.compile(optimizer='adam', loss='mse')
        print("Modelo LSTM construído e compilado.")
        self.lstm_model.summary()

        # Callbacks
        early_stopping = EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True)
        reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=patience // 2, min_lr=0.0001)

        # Train Model
        history = self.lstm_model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size,
                                      validation_data=(X_test, y_test), callbacks=[early_stopping, reduce_lr], verbose=1)
        print("Ancoragem híbrida LSTM concluída.")

        # Evaluate and store results
        y_pred_scaled = self.lstm_model.predict(X_test)
        # Inverse transform to get actual values
        # Create dummy arrays for features not being predicted, set to average or 0
        y_test_full = np.zeros_like(scaled_data[-len(y_test):])
        y_pred_full = np.zeros_like(scaled_data[-len(y_pred_scaled):])

        # Place actual target variable predictions/true values into the correct column
        target_idx = data_for_lstm.columns.get_loc(target_variable)
        y_test_full[:, target_idx] = y_test[:, data_for_lstm.columns.get_loc(target_variable)]
        y_pred_full[:, target_idx] = y_pred_scaled[:, 0] # Assuming single target prediction

        y_test_inv = self.scaler.inverse_transform(y_test_full)[:, target_idx]
        y_pred_inv = self.scaler.inverse_transform(y_pred_full)[:, target_idx]

        rmse = np.sqrt(mean_squared_error(y_test_inv, y_pred_inv))
        self.model_results['lstm_hybrid_anchoring'] = {'rmse': rmse, 'history': history.history,
                                                      'y_test': y_test_inv, 'y_pred': y_pred_inv}
        print(f"LSTM RMSE no conjunto de teste: {rmse:.4f}")
        return self.model_results['lstm_hybrid_anchoring']

    def decomposicao_variancia_fevd(self, var_results, horizon=10):
        print(f"Calculando Decomposição da Variância de Erro de Previsão (FEVD) para horizonte {horizon}...")
        if var_results is None:
            print("Resultados VAR não fornecidos para FEVD.")
            return None

        fevd = var_results.fevd(horizon=horizon)
        self.model_results['fevd'] = fevd.summary()
        print("FEVD calculado.")
        return fevd.summary()

    def gerar_e_exportar_irf(self, var_results, periods=10, orth=True, export_path=None):
        print(f"Gerando e exportando Funções de Resposta a Impulso (IRF) para {periods} períodos...")
        if var_results is None:
            print("Resultados VAR não fornecidos para IRF.")
            return None

        irf = var_results.irf(periods=periods)
        self.model_results['irf'] = {'irf_object': irf, 'summary': irf.summary()}

        if export_path:
            # Example: export to Excel
            with pd.ExcelWriter(export_path) as writer:
                for shock in var_results.names:
                    for response in var_results.names:
                        irf.plot(impulse=shock, response=response, orth=orth)
                        # In a notebook, this would display plots. For export, you might save figures or data.
                        # For simplicity, let's just save the raw IRF data for each impulse-response pair.
                        irf_data = pd.DataFrame({
                            'Periods': range(periods + 1),
                            'IRF_Value': irf.irfs[shock][:, var_results.names.index(response)]
                        })
                        irf_data.to_excel(writer, sheet_name=f'IRF_{shock}_to_{response}', index=False)
            print(f"IRFs exportadas para {export_path}")

        print("IRFs geradas e processadas.")
        return irf

    def carregar_dados_csv_personalizados(self, file_path, date_column=None, index_col=None):
        print(f"Carregando dados CSV personalizados de {file_path}...")
        loader = DataLoader(file_path=file_path)
        df = loader.load_csv()
        if df is not None:
            if date_column and date_column in df.columns:
                df[date_column] = pd.to_datetime(df[date_column])
                df = df.set_index(date_column)
                print(f"Coluna '{date_column}' definida como índice de data.")
            elif index_col:
                df = df.set_index(index_col)
                print(f"Coluna '{index_col}' definida como índice.")
            self.df = df
            print("Dados CSV personalizados carregados e processados.")
        else:
            print("Falha ao carregar dados CSV personalizados.")
        return self.df

    def run_full_pipeline(self):
        print("\n--- Executando Pipeline Completo do MacroLab Engine --- ")

        # 1. Ingest Data
        self.ingest_data()
        if self.df is None or self.df.empty:
            print("Pipeline interrompido: Nenhuma dado disponível após ingestão.")
            return
        print(f"DataFrame após ingestão (primeiras 5 linhas):\n{self.df.head()}\n")

        # 2. Perform Unit Root Tests
        if self.features:
            unit_root_results = self.perform_unit_root_tests(columns=self.features + [self.target_variable] if self.target_variable else self.features)
            if unit_root_results: print("Resultados dos testes de raiz unitária:\n", unit_root_results, "\n")

        # 3. Estimate SPJ (PanelOLS) if applicable
        if self.country_col and self.time_col and self.target_variable and self.features:
            spj_results = self.estimate_spj(self.target_variable, self.features)
            if spj_results: print("Resultados SPJ disponíveis no self.model_results['spj_estimator']\n")

        # 4. Estimate VAR/VECM if applicable
        if self.features and len(self.features) > 1:
            var_vecm_results = self.estimate_var_vecm(self.features)
            if var_vecm_results: print("Resultados VAR/VECM disponíveis no self.model_results['var_vecm']\n")

            # 5. FEVD and IRF (require VAR results)
            if 'var_vecm' in self.model_results:
                fevd_summary = self.decomposicao_variancia_fevd(var_vecm_results)
                if fevd_summary: print("Resumo FEVD disponível no self.model_results['fevd']\n")

                irf_export_path = os.path.join(self.project_root, 'resultados', 'irf_results.xlsx')
                os.makedirs(os.path.dirname(irf_export_path), exist_ok=True)
                irf_object = self.gerar_e_exportar_irf(var_vecm_results, export_path=irf_export_path)
                if irf_object: print("IRF exportado e disponível no self.model_results['irf']\n")

        # 6. Kink Fiscal Mapping (example usage, assumes relevant columns are set up)
        # This part is highly context-specific and requires specific column names.
        # Example: if 'tax_rate', 'threshold_income', 'total_income' exist in self.df
        # self.mapeamento_kink_fiscal('tax_rate', 'threshold_income', 'total_income', tax_elasticity=0.1)
        print("Ignorando mapeamento de kink fiscal para execução completa, pois requer colunas específicas.\n")

        # 7. Hybrid LSTM Anchoring (requires target_variable and features)
        if self.target_variable and self.features:
            lstm_results = self.ancoragem_hibrida_lstm(self.target_variable, self.features)
            if lstm_results: print("Resultados LSTM disponíveis no self.model_results['lstm_hybrid_anchoring']\n")

        print("--- Pipeline Completo do MacroLab Engine Concluído --- ")
        return self.model_results

"""

macro_engine_file_path = os.path.join(project_root, 'macrolab', 'macro_engine.py')

with open(macro_engine_file_path, 'w') as f:
    f.write(macro_engine_content)

print(f"Content written to {macro_engine_file_path}")
````

````python
import os

project_root = '/content/drive/MyDrive/meu_projeto_tese'

# Detailed content for macrolab/econometrics.py
econometrics_detailed_content = """
# macrolab/econometrics.py

import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller, kpss
from statsmodels.tsa.vector_ar.var_model import VAR
from statsmodels.tsa.vector_ar.vecm import VECM, select_order, coint_johansen
from linearmodels.panel import PanelOLS

class SPJEstimator:
    '''
    Implementação de um Estimador SPJ, utilizando PanelOLS para dados em painel.
    Esta classe é adequada para análises que requerem efeitos fixos ou aleatórios.
    '''
    def __init__(self, data_panel=None, dependent_var=None, independent_vars=None,
                 entity_effects=True, time_effects=True, formula=None):
        self.data_panel = data_panel # Deve ser um DataFrame MultiIndex (entity, time)
        self.dependent_var = dependent_var
        self.independent_vars = independent_vars
        self.entity_effects = entity_effects
        self.time_effects = time_effects
        self.formula = formula
        self.model = None
        self.results = None

    def fit(self):
        if self.data_panel is None or self.data_panel.empty:
            print("Erro: DataFrame do painel vazio ou não fornecido para SPJ Estimator.")
            return None
        if self.dependent_var is None or self.independent_vars is None:
            print("Erro: Variáveis dependente ou independentes não especificadas para SPJ Estimator.")
            return None

        print(f"Fitting SPJ Estimator (PanelOLS) for {self.dependent_var} ~ {', '.join(self.independent_vars)}")
        try:
            dependent = self.data_panel[self.dependent_var]
            exog = self.data_panel[self.independent_vars]
            exog = sm.add_constant(exog, prepend=False) # Adicionar constante

            self.model = PanelOLS(dependent, exog, entity_effects=self.entity_effects, time_effects=self.time_effects)
            self.results = self.model.fit()
            print("SPJ Estimator fitted successfully.")
            print(self.results)
            return self.results
        except Exception as e:
            print(f"Erro ao ajustar SPJ Estimator: {e}")
            return None

class VARVECM:
    '''
    Implementação de modelos VAR (Vector Autoregression) e VECM (Vector Error Correction Model).
    Detecta a ordem do VAR e testa a co-integração para decidir entre VAR e VECM.
    '''
    def __init__(self, data=None, max_lags=5, significance=0.05):
        self.data = data # Deve ser um DataFrame de séries temporais
        self.max_lags = max_lags
        self.significance = significance
        self.model_type = None
        self.model = None
        self.results = None
        self.optimal_lag = None

    def _select_var_order(self):
        if self.data is None or self.data.empty:
            return None
        print(f"Selecionando a ordem ótima do VAR (max_lags={self.max_lags})...")
        # using default criteria, could be customized (e.g., ic='aic')
        model = VAR(self.data)
        selected_order = select_order(self.data, maxlags=self.max_lags, verbose=False)
        self.optimal_lag = selected_order.aic # Use AIC for optimal lag
        print(f"Ordem ótima do VAR selecionada (AIC): {self.optimal_lag}")
        return self.optimal_lag

    def _test_cointegration(self):
        if self.data is None or self.data.empty:
            return False
        print("Testando co-integração (teste de Johansen)...")
        try:
            # Determine VAR order first for coint_johansen
            if self.optimal_lag is None:
                self._select_var_order()
            if self.optimal_lag is None: # If selection failed
                print("Não foi possível selecionar a ordem do VAR, pulando o teste de co-integração.")
                return False

            # k_ar_diff is the VAR order of the differenced series, typically optimal_lag-1 for VECM
            cvf = coint_johansen(self.data, det_order=0, k_ar_diff=self.optimal_lag - 1)
            # Check trace statistic against critical values (e.g., 95% critical value)
            # If the largest eigenvalue is greater than critical value, there's at least 1 cointegrating relationship
            if cvf.trace_stat[0] > cvf.trace_crit[0, 1]: # Trace statistic for r=0 vs r=1 (95% critical value)
                print(f"Co-integração detectada (r={cvf.trace_stat.size - np.where(cvf.trace_stat > cvf.trace_crit[:, 1])[0][-1] if cvf.trace_stat.size > 0 else 0} relações).")
                return True
            else:
                print("Nenhuma co-integração detectada.")
                return False
        except Exception as e:
            print(f"Erro no teste de co-integração: {e}")
            return False

    def fit(self):
        if self.data is None or self.data.empty:
            print("Nenhum dado fornecido para o modelo VAR/VECM.")
            return None

        data_cleaned = self.data.dropna() # Remover NaNs para modelagem de séries temporais
        if data_cleaned.empty:
            print("Dados vazios após remover NaNs.")
            return None

        self._select_var_order()
        if self.optimal_lag is None: # Fallback if order selection fails
            self.optimal_lag = 1
            print(f"Utilizando lag padrão de {self.optimal_lag} devido a falha na seleção da ordem.")

        # Test cointegration to decide between VAR and VECM
        is_cointegrated = self._test_cointegration()

        if is_cointegrated:
            self.model_type = 'VECM'
            print(f"Ajustando VECM com {self.optimal_lag-1} defasagens e det_order=0...") # k_ar_diff is VAR order - 1
            try:
                # VECM needs VAR order of differenced data (k_ar_diff)
                self.model = VECM(data_cleaned, k_ar_diff=self.optimal_lag - 1, coint_rank=1, dates=data_cleaned.index) # coint_rank needs to be determined programmatically or specified
                self.results = self.model.fit()
                print("Modelo VECM ajustado com sucesso.")
            except Exception as e:
                print(f"Erro ao ajustar VECM: {e}")
                print("Caindo para modelo VAR.")
                self.model_type = 'VAR'
        else:
            self.model_type = 'VAR'

        if self.model_type == 'VAR':
            print(f"Ajustando VAR com {self.optimal_lag} defasagens...")
            try:
                self.model = VAR(data_cleaned)
                self.results = self.model.fit(self.optimal_lag)
                print("Modelo VAR ajustado com sucesso.")
            except Exception as e:
                print(f"Erro ao ajustar VAR: {e}")
                return None

        if self.results:
            print(self.results.summary())
        return self.results

def adf_test(series, name=''):
    '''
    Realiza um teste de Dickey-Fuller Aumentado (ADF) em uma série temporal.
    Retorna o p-valor e a conclusão sobre a estacionariedade.
    '''
    print(f"Realizando teste ADF para a série: {name}")
    if series.isnull().any():
        print(f"A série '{name}' contém valores ausentes. Removendo-os para o teste ADF.")
        series = series.dropna()
    if series.empty or len(series) < 5: # Mínimo de observações para o teste
        print(f"A série '{name}' é muito curta ou vazia para o teste ADF.")
        return {
            'ADF_Statistic': np.nan,
            'P-value': np.nan,
            'Lags Used': np.nan,
            'No. of Observations': np.nan,
            'Critical Values': {},
            'Conclusion': 'Não foi possível realizar o teste: dados insuficientes.'
        }

    try:
        adf_result = adfuller(series, autolag='AIC')
        adf_pvalue = adf_result[1]
        conclusion = 'Estacionária (rejeita H0)' if adf_pvalue <= 0.05 else 'Não Estacionária (falha ao rejeitar H0)'

        results = {
            'ADF_Statistic': adf_result[0],
            'P-value': adf_pvalue,
            'Lags Used': adf_result[2],
            'No. of Observations': adf_result[3],
            'Critical Values': adf_result[4],
            'Conclusion': conclusion
        }
        print(f"  P-value: {adf_pvalue:.4f} -> {conclusion}")
        return results
    except Exception as e:
        print(f"Erro ao executar o teste ADF para '{name}': {e}")
        return {
            'ADF_Statistic': np.nan,
            'P-value': np.nan,
            'Lags Used': np.nan,
            'No. of Observations': np.nan,
            'Critical Values': {},
            'Conclusion': f'Erro ao realizar o teste: {e}'
        }

def kpss_test(series, name=''):
    '''
    Realiza um teste KPSS em uma série temporal.
    A hipótese nula do KPSS é que a série é estacionária.
    '''
    print(f"Realizando teste KPSS para a série: {name}")
    if series.isnull().any():
        print(f"A série '{name}' contém valores ausentes. Removendo-os para o teste KPSS.")
        series = series.dropna()
    if series.empty or len(series) < 5:
        print(f"A série '{name}' é muito curta ou vazia para o teste KPSS.")
        return {
            'KPSS_Statistic': np.nan,
            'P-value': np.nan,
            'Lags Used': np.nan,
            'No. of Observations': np.nan,
            'Critical Values': {},
            'Conclusion': 'Não foi possível realizar o teste: dados insuficientes.'
        }

    try:
        kpss_result = kpss(series, regression='c', nlags='auto')
        kpss_pvalue = kpss_result[1]
        conclusion = 'Não Estacionária (rejeita H0)' if kpss_pvalue <= 0.05 else 'Estacionária (falha ao rejeitar H0)'

        results = {
            'KPSS_Statistic': kpss_result[0],
            'P-value': kpss_pvalue,
            'Lags Used': kpss_result[2],
            'Critical Values': kpss_result[3],
            'Conclusion': conclusion
        }
        print(f"  P-value: {kpss_pvalue:.4f} -> {conclusion}")
        return results
    except Exception as e:
        print(f"Erro ao executar o teste KPSS para '{name}': {e}")
        return {
            'KPSS_Statistic': np.nan,
            'P-value': np.nan,
            'Lags Used': np.nan,
            'Critical Values': {},
            'Conclusion': f'Erro ao realizar o teste: {e}'
        }

"""

econometrics_file_path = os.path.join(project_root, 'macrolab', 'econometrics.py')

with open(econometrics_file_path, 'w') as f:
    f.write(econometrics_detailed_content)

print(f"Conteúdo detalhado escrito para {econometrics_file_path}")
````
